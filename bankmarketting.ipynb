*BANK MARKETING*

The Bank Marketing Dataset contains information collected from direct marketing campaigns conducted by a bank. The campaigns were primarily carried out through telephone calls to encourage customers to subscribe to a term deposit. The dataset includes customer demographic information, previous campaign interactions, loan details, and economic indicators that may influence customer decisions.

This dataset consists of 39,188 customer records and 21 variables, making it suitable for data cleaning, exploratory data analysis (EDA), data visualisation, and predictive analytics projects.


**Business Objective**

The primary objective of this project is to analyse customer and campaign-related factors that influence whether a customer subscribes to a bank term deposit.

By understanding these factors, the bank can:

Identify potential customers who are more likely to subscribe.
Improve the effectiveness of future marketing campaigns.
Reduce marketing costs by targeting the right customer segments.
Increase the success rate of term deposit subscriptions.

```python
import seaborn as sns
import matplotlib.pyplot as plt

outlier_cols = ['age','duration','campaign','pdays','previous','cons.conf.idx']

for col in outlier_cols:
    plt.figure(figsize=(10,4))
    sns.boxplot(x=df[col], color='skyblue')

    plt.title(f'Boxplot Analysis of {col.capitalize()}', fontsize=14)
    plt.xlabel(col.capitalize(), fontsize=12)
    plt.ylabel('Values', fontsize=12)

    plt.show()

    print(f"Analysis: The boxplot of {col} helps identify potential outliers and the spread of data.")
```

Analysis Notes Below Each Boxplot
Age

Analysis: Most customers are concentrated between the ages of 30 and 50 years. A few extreme age values are present, indicating potential outliers among very young or older customers.

Duration

Analysis: Call duration exhibits a large number of outliers, indicating some calls lasted significantly longer than average. Customers with longer call durations tend to subscribe more frequently. However, duration is only known after the call is completed, so it cannot be used for pre-call targeting decisions.

Campaign

Analysis: Most customers were contacted only a few times, while a small number received many contacts. Excessive contact attempts appear to have diminishing returns and may reduce campaign effectiveness.

Pdays

Analysis: The distribution shows many extreme values because a large proportion of customers had not been contacted previously. This indicates that many campaign contacts were first-time interactions.

Previous

Analysis: Most customers had few or no previous contacts. Customers with successful prior interactions may show higher subscription rates.

Consumer Confidence Index (cons.conf.idx)

Analysis: The data contains some outliers, but the overall variation is moderate. Consumer confidence may influence customers' willingness to subscribe to financial products.

Check Outliers Using IQR Method

```python
Q1 = df['age'].quantile(0.25)
Q3 = df['age'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['age'] < lower) | (df['age'] > upper)]

print("Number of outliers:", len(outliers))
```

Outlier Detection in Age

Method Used:
 Outliers in the Age variable were identified using the Interquartile Range (IQR) method.

Q1 (25th percentile) and Q3 (75th percentile) were calculated.
IQR was computed as:
IQR=Q3−Q1IQR = Q3 - Q1IQR=Q3−Q1
Lower and upper bounds were determined using:
Lower Bound=Q1−1.5×IQR\text{Lower Bound} = Q1 - 1.5 \times IQRLower Bound=Q1−1.5×IQR Upper Bound=Q3+1.5×IQR\text{Upper Bound} = Q3 + 1.5 \times IQRUpper Bound=Q3+1.5×IQR

Result:
 A total of 438 outlier observations were detected in the Age feature.

Interpretation:
 The presence of 438 outliers indicates that some customers are significantly younger or older than the majority of customers in the dataset. Since age is an important customer characteristic and these values may represent valid customer records rather than data entry errors, the outliers were retained for further analysis.

Subscription Distribution

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='y', data=df)

plt.title('Distribution of Term Deposit Subscription', fontsize=14)
plt.xlabel('Subscription Status', fontsize=12)
plt.ylabel('Number of Customers', fontsize=12)
plt.show()
```

Analysis:
 The distribution shows that the majority of customers did not subscribe to the term deposit, while a smaller proportion subscribed. This indicates a class imbalance in the target variable, which should be considered during model building and evaluation. From a business perspective, understanding the characteristics of the subscribed customers can help improve future marketing campaign effectiveness

**Month Analysis**

```python
month_order = ['mar', 'apr', 'may', 'jun', 'jul',
               'aug', 'sep', 'oct', 'nov', 'dec']

month_percent = pd.crosstab(
    df['month'],
    df['y'],
    normalize='index'
) * 100

month_percent = month_percent.reindex(month_order)

month_percent['yes'].plot(
    kind='bar',
    figsize=(10,5),
    color='green'
)

plt.title('Monthly Subscription Rate (%)')
plt.xlabel('Month')
plt.ylabel('Subscription Rate (%)')

plt.show()
```

These insights can help the bank schedule future marketing campaigns during periods with stronger subscription performance.
